### Task 1

Suppose you have a list named my_list as given below. <br/>
my_list = [10,1,20,3,6,2,5,11,15,2,12,14,17,18,29] <br/>
Now use bubble sort to sort my_list into ascending order.

**Output**<br/>
[1, 2, 2, 3, 5, 6, 10, 11, 12, 14, 15, 17, 18, 20, 29]

```python
#todo - Bubble sort starts arranging the elements from the top
my_list = [10,1,20,3,6,2,5,11,15,2,12,14,17,18,29]
for idx1 in range(len(my_list)-1):
    for idx2 in range(len(my_list)-idx1-1):
        if my_list[idx2]>my_list[idx2+1]:
            temp=my_list[idx2]
            my_list[idx2]=my_list[idx2+1]
            my_list[idx2+1]=temp
print(my_list)
```

### Task 2

Suppose you have a list named my_list as given below. <br/>
my_list = [10,1,20,3,6,2,5,11,15,2,12,14,17,18,29] <br/>
Now use selection sort to sort my_list into ascending order.

**Output**<br/>
[1, 2, 2, 3, 5, 6, 10, 11, 12, 14, 15, 17, 18, 20, 29]



```python
#todo Selection sort starts arranging the elements from the bottom
my_list = [10,1,20,3,6,2,5,11,15,2,12,14,17,18,29]
def selection_sorting(arr):
    for idx1 in range(len(arr)):
        min_val = arr[idx1]
        min_idx = idx1
        for idx2 in range(idx1+1,len(arr)):
            if arr[idx2]<min_val:
                min_val=arr[idx2]
                min_idx=idx2
        temp=arr[min_idx]
        arr[min_idx]=arr[idx1]
        arr[idx1]=temp
    return arr
sorted_num=selection_sorting(my_list)
print(sorted_num)
```

```python
sitting_list = [10,30,20,70,11,15,22,16,58,100,12,56,70,80]
def selection_sorting(arr):
    for idx1 in range(len(arr)):
        if idx1%2!=0:
            max_val = arr[idx1]
            max_idx = idx1
            for idx2 in range(idx1+2,len(arr),2):
                if arr[idx2]>max_val:
                    max_val=arr[idx2]
                    max_idx=idx2
            temp=arr[max_idx]
            arr[max_idx]=arr[idx1]
            arr[idx1]=temp
        else:
            min_val = arr[idx1]
            min_idx = idx1
            for idx3 in range(idx1+2,len(arr),2):
                if arr[idx3]<min_val:
                    min_val=arr[idx3]
                    min_idx=idx3
            temp1=arr[min_idx]
            arr[min_idx]=arr[idx1]
            arr[idx1]=temp1
    return arr
sorted_num=selection_sorting(sitting_list)
print(sorted_num)
```

```python
list_one = [1, 2, 1, 4]
list_two = [5, 4, 1]
n_list=list_one+list_two
n_list.sort()
print('Sorted list =',n_list)
if len(n_list)%2==0:
    print('Median =',((n_list[len(n_list)//2-1]+n_list[(len(n_list)//2)])/2))
else:
     print('Median =',n_list[len(n_list)//2])
```
