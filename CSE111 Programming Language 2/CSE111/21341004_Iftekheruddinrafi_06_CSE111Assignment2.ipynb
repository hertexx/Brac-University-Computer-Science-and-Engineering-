```python
#Q1 Answer
def fun(val1,val2):
    try:
        rawdiv=val1/val2
        tempdiv=val1//val2
        result=rawdiv-tempdiv
        return result
    except ZeroDivisionError:
        return 0
fun(5,0)
```

```python
#Q2 Answer
def fun(height,weight):
    bmi=weight/(height/100)**2
    if bmi<18.5:
        print("Score is "+str(round(bmi,1))+". You are Underweight")
    elif 18.5<=bmi<24.9:
        print("Score is "+str(round(bmi,1))+". You are Normal")
    elif 25<=bmi<=30:
        print("Score is "+str(round(bmi,1))+". You are Overweight")
    elif bmi>30:
        print("Score is "+str(round(bmi,1))+". You are Obese")
fun(175,96)
fun(152,48)
```

```python
#Q3 Answer
def fun(minm,maxm,div):
    total=0
    for item in range(minm,maxm):
        if item%div==0:
            total+=item
    return total
fun(3, 16, 3)
```

```python
#Q6 Answer
name=input('Enter your full name: ').lower()
def fun(name):
    vowl_cnt=0
    vowels=['a','e','i','o','u']
    vowl_present=[]
    for char in name:
        if char in vowels:
            vowl_cnt+=1
            vowl_present.append(char)
    if vowl_cnt>=1:
        print('Vowels:',end=' ')
        for alph in range(len(vowl_present)):
            if alph==len(vowl_present)-1:
                print(vowl_present[alph],end='.')
            else:
                print(vowl_present[alph],end=',')
        print(' Total number of vowels:',vowl_cnt)
    else:
        print('No vowels in name')
fun(name)
```

```python
#Q7 Answer
def fun(word):
    strg1=''
    strg2=''
    for item in word:
        if 97<=ord(item)<=122:
            strg1+=item
    for char in range(-1,-(len(word)+1),-1):
        if 97<=ord(word[char])<=122:
            strg2+=word[char]
    if strg1==strg2:
        print('Palindrome')
    else:
        print('Not a palindrome')
fun('nurses run')
```

```python
#Q9 Answer
def fun(strg):
    string='  '+strg
    lst=[]
    newstr=''
    cnt=0
    for item in range(len(string)):
        if '.' == string[item] or '!' == string[item] or '?' == string[item]:
            lst.append(string[cnt+1:item+1])
            cnt=item
    for elem in range(len(lst)):
        for alpha in range(len(lst[elem])):
            if alpha==1:
                newstr+=lst[elem][alpha].upper()
            else:
                newstr+=lst[elem][alpha]
    return newstr.lstrip()
print(fun('my favourite animal is a dog. a dog has sharp teeth so that it can eat flesh very easily. do you know my pet dog’s name? i love my pet very much.'))
```

```python
#Q9 Answer
def fun(strg1):
    myDic={}
    strg=strg1+'  '
    strg2=''
    for elem in range(len(strg)):
        if elem==0:
            myDic[elem]=strg[elem].upper()
        elif strg[elem]=='.' or strg[elem]=='?' or strg[elem]=='!':
            myDic[elem]=strg[elem+2].upper()
    for char in range(len(strg)):
        if char in myDic.keys():
            if char==0:
                strg2+=strg[char]
                strg2+=myDic.get(char)
        else:
            strg2+=strg[char]
    return strg2
print(fun('my favourite animal is a dog. a dog has sharp teeth so that it can eat flesh very easily. do you know my pet dog’s name? i love my pet very much.'))
```
