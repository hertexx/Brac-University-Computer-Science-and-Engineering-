```python
#Q4 Answer
lst=[]
while True:
    user_input= [element for element in input("Enter the list : ").translate(dict.fromkeys([ord('['),ord(']')])).split(' ')]
    if user_input==['STOP']:
        break
    else:
        lst.append(user_input)
ub_lst=[]
ub_checker=True
for item in range(len(lst)):
    for items in range(len(lst[item])-1):
        ub_lst.append(abs(int(lst[item][items+1])-int(lst[item][items])))
    for num in range(1,len(lst[item])):
        if num not in ub_lst:
            ub_checker=False    
    if ub_checker==True:
        print('UB Jumper')
    else:
        print('Not UB Jumper')
    ub_checker=True
    ub_lst.clear()
```

# Dictionary & Tuple

```python
#Q1 Answer
user_inp1=[element for element in input("").translate(dict.fromkeys([ord('['),ord(']')])).split(',')]
user_inp2=[element for element in input("").translate(dict.fromkeys([ord('['),ord(']')])).split(',')]
dictionary_1 = {}
dictionary_2 = {}
for x in user_inp1:
    k,v = x.split(':')
    dictionary_1[k.strip()] = int(v)
for y in user_inp2:
    k,v = y.split(':')
    dictionary_2[k.strip()] = int(v)
dictionary_3 = dict(dictionary_1)
dictionary_3.update(dictionary_2)
unique_values = []
for dict_1_key, dict_1_value in dictionary_1.items():
    for dict_2_key, dict_2_value in dictionary_2.items():
        if (dict_1_key == dict_2_key):
            dictionary_3[dict_1_key]=(dict_1_value + dict_2_value)
print(dictionary_3)
for value in dictionary_3.values(): 
    if value in unique_values: 
        continue 
    else:
        unique_values.append(value)
print(tuple(sorted(unique_values)))
```
