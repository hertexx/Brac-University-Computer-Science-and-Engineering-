```python
import sympy as smp
a, b , x = smp.symbols("a b x")
f= smp.sin(a*x)*smp.exp(-b*x)
smp.integrate(f, x).evalf()
```

```python
import numpy as np
import scipy as sp
from scipy.integrate import quad 
f= lambda x: (np.sin(2*x)**-1)/  1+x**2
quad(f, 0,np.inf).evalf()
```

```python
import sympy as smp
import math
a, b , x = smp.symbols("a b x")
f = 1/ ((a-smp.cos(x))**2) + ((b-smp.sin(x))**2)
smp.integrate(f, (x,0, 2*math.pi)).evalf()
```

```python
import sympy as smp
import math
a, b , x = smp.symbols("a b x")
f = 1/ ((a-smp.cos(x))**2) + ((b-smp.sin(x))**2)
smp.integrate(f, (x,0, 2*math.pi)).simplify().evalf()
```
