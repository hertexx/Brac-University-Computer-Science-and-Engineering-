```python
#Q1
class Calculator:
    def __init__(self,num1,orp,num2):
        self.num1=num1
        self.orp=str(orp)
        self.num2=num2
        print("Let's Calculate!")
    def add(self):
        print('Result:',self.num1+self.num2)
    def subtract(self):
        print('Result:',self.num1-self.num2)
    def multiply(self):
        print('Result:',self.num1*self.num2)
    def divide(self):
        print('Result:',self.num1/self.num2)      
val1=int(input(''))
opt=input('')
val2=int(input(''))
cal=Calculator(val1,opt,val2)
print('Value 1:',val1)
print('Operator:',opt)
print('Value 2:',val2)
if opt=='+':
    cal.add()
elif opt=='-':
    cal.subtract()
elif opt=='*':
    cal.multiply()
elif opt =='/':
    cal.divide()
```

```python
#Q2
class Customer:
    def __init__(self,name):
        self.name=name
    def greet(self, gname=''):
        if len(gname)>0:
            print('Hello '+gname+'!')
        else:
            print('Hello'+gname+'!')
    def purchase(self,*item):
        print(self.name+', you have purshased '+str(len(item)),'item(s)')
        for product in item:
            print(product)
customer_1 = Customer("Sam") 
customer_1.greet() 
customer_1.purchase("chips", "chocolate", "orange juice")
print("-----------------------------") 
customer_2 = Customer("David") 
customer_2.greet("David") 
customer_2.purchase("orange juice")
```

```python
#Q3
class Panda:
    def __init__(self,name,gender,age):
        self.name=name
        self.gender=gender
        self.age=age
    def sleep(self,nap=0):
        self.nap=nap
        if 3<=nap<=5:
            return f"{self.name} sleeps {self.nap} hours daily and should have  Mixed Veggies"
        elif 6<=nap<=8:
            return f"{self.name} sleeps {self.nap} hours daily and should have  Eggplant & Tofu"
        elif 9<=nap<=11:
            return f"{self.name} sleeps {self.nap} hours daily and should have  Broccoli Chicken"
        else:
            return f"{self.name} duration is unknown thus should  have only bamboo leaves"
panda1 = Panda("Kunfu","Male", 5) 
panda2=Panda("Pan Pan","Female",3) 
panda3=Panda("Ming Ming","Female",8) 
print("{} is a {} Panda Bear who is {} years old".format(panda1.name,panda1.gender,panda1.age)) 
print("{} is a {} Panda Bear who is {} years old".format(panda2.name,panda2.gender,panda2.age)) 
print("{} is a {} Panda Bear who is {} years old".format(panda3.name,panda3.gender,panda3.age))
print("===========================")
print(panda2.sleep(10))
print(panda1.sleep(4))
print(panda3.sleep())
```

```python
#Q5
class Student:
    
    def __init__(self,name='default student'):
        self.name=name
    def quizcalc(self,*score):
        self.avg=0
        tot=0
        for item in score:
            tot+=item
        self.avg=tot/3
    def printdetail(self):
        print('Hello',self.name)
        print('Your average quiz score is: ',self.avg)
s1 = Student() 
s1.quizcalc(10) 
print('--------------------------------')
s1.printdetail() 
s2 = Student('Harry')
s2.quizcalc(10,8) 
print('--------------------------------')
s2.printdetail() 
s3 = Student('Hermione') 
s3.quizcalc(10,9,10) 
print('--------------------------------')
s3.printdetail()
```

```python
#Q9
class Patient:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def add_Symptom(self,*illness):
        if len(illness)==3:
            self.illness1=illness[0]
            self.illness2=', '+illness[1]
            self.illness3=', '+illness[2]
        elif len(illness)==2:
            self.illness1=illness[0]
            self.illness2=', '+illness[1]
            self.illness3=''
        elif len(illness)==1:
            self.illness1=illness[0]
            self.illness2=''
            self.illness3=''
    def printPatientDetail(self):
        print('Name:',self.name)
        print('Age:',self.age)
        print('Symptoms: '+self.illness1+self.illness2+self.illness3)
p1 = Patient("Thomas", 23) 
p1.add_Symptom("Headache") 
p2 = Patient("Carol", 20) 
p2.add_Symptom("Vomiting", "Coughing") 
p3 = Patient("Mike", 25) 
p3.add_Symptom("Fever", "Headache", "Coughing")
print("=========================") 
p1.printPatientDetail() 
print("=========================") 
p2.printPatientDetail() 
print("=========================") 
p3.printPatientDetail() 
print("=========================")
```

```python
#Q11
class Shinobi:
    def __init__(self,name,rank,salary=0,miss=0):
        self.name=name
        self.rank=rank
        self.miss=miss
        if self.rank=='Genin':
            self.salary=250
        elif self.rank=='Chunin':
            self.salary=1000
        elif self.rank=='Jonin':
            self.salary=2500
    def changeRank(self,rank):
        self.rank=rank
    def calSalary(self,miss):
        self.miss=miss
    def printInfo(self):
        if self.miss>0:
            print('Name:', self.name)
            print('Rank:', self.rank)
            print('Number of mission:',self.miss)
            print('Salary:',self.salary)
        else:
            print('Name:', self.name)
            print('Rank:', self.rank)
            print('Number of mission:',self.miss)
            print('Salary: 0')
naruto = Shinobi("Naruto", "Genin") 
naruto.calSalary(5) 
naruto.printInfo() 
print('====================') 
shikamaru = Shinobi('Shikamaru', "Genin")
shikamaru.printInfo() 
shikamaru.changeRank("Chunin") 
shikamaru.calSalary(10) 
shikamaru.printInfo() 
print('====================') 
neiji = Shinobi("Neiji", "Jonin") 
neiji.calSalary(5) 
neiji.printInfo()
```

```python
#Q13
class Batsman:
    def __init__(self,*info):
        if len(info)==2:
            self.name='New Batsman'
            self.score=info[0]
            self.face=info[1]
        elif len(info)==3:
            self.name=info[0]
            self.score=info[1]
            self.face=info[2]
    def printCareerStatistics(self):
        print('Name:',self.name)
        print('Run Scored:',self.score,', Ball Faced:',self.face)
    def setName(self,name):
        self.name=name
    def battingStrikeRate(self):
        return (self.score/self.face)*100
b1 = Batsman(6101, 7380) 
b1.printCareerStatistics() 
print("============================")
b2 = Batsman("Liton Das", 678, 773)
b2.printCareerStatistics() 
print("----------------------------") 
print(b2.battingStrikeRate()) 
print("============================")
b1.setName("Shakib Al Hasan") 
b1.printCareerStatistics() 
print("----------------------------") 
print(b1.battingStrikeRate())
```

```python
#Q15
class Account:
    def __init__(self,name='Default Account',balance=0):
        self.name=name
        self.balance=balance
    def name(self,name):
        self.name=name
    def balance(self,balance):
        self.balance=balance
    def withdraw(self,balance):
        if self.balance-balance<=3070:
            print('Sorry, Withdraw unsuccessful! The account  balance after deducting withdraw amount is  equal to or less than minimum.')
        else:
            self.balance-=balance
            print('Withdraw successful! New balance is:',self.balance)
    def details(self):
        return f"{self.name}\n{float(self.balance)}"
a1 = Account() 
print(a1.details()) 
print("------------------------") 
a1.name = "Oliver" 
a1.balance = 10000.0 
print(a1.details()) 
print("------------------------") 
a2 = Account("Liam") 
print(a2.details()) 
print("------------------------") 
a3 = Account("Noah",400) 
print(a3.details()) 
print("------------------------") 
a1.withdraw(6930); 
print("------------------------") 
a2.withdraw(600); 
print("------------------------") 
a1.withdraw(6929)
```

```python
#Q16
class Author:
    def __init__(self,*info):
        self.name='Default'
        if len(info)==1:
            self.name=info[0]
        elif len(info)==3:
            self.name=info[0]
            self.book1=info[1]
            self.book2=info[2]
            self.book3=''
    def addBooks(self,*book):
        if len(book)==2:
            self.book1=book[0]
            self.book2=book[1]
            self.book3=''
        elif len(book)==3:
            self.book1=book[0]
            self.book2=book[1]
            self.book3=book[2]
    def changeName(self,name):
        self.name=name
    def printDetails(self):
        print('Auther Name:',self.name)
        print('--------')
        print(f"List of Books:\n{self.book1}\n{self.book2}\n{self.book3}")
auth1 = Author('Humayun Ahmed') 
auth1.addBooks('Deyal', 'Megher Opor Bari') 
auth1.printDetails() 
print('===================') 
auth2 = Author() 
print(auth2.name) 
auth2.changeName('Mario Puzo') 
auth2.addBooks('The Godfather', 'Omerta', 'The Sicilian')
print('===================') 
auth2.printDetails() 
print('===================') 
auth3 = Author('Paolo Coelho', 'The Alchemist', 'The Fifth Mountain')
auth3.printDetails()
```
