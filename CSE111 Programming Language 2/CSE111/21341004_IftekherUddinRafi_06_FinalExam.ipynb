```python
# AnswerQ1
lst=[]
divlst=[]
digtsum=0
plst=[]
implst=[]
while True:
    usrinp=input('')
    if usrinp=='STOP':
        break
    else:
        lst.append(int(usrinp))
for item in lst:
    for num in range(1,item):
        if item%num==0:
            divlst.append(num)
    for elem in divlst:
        digtsum+=elem
    if digtsum==item:
        plst.append(item)
    else:
        implst.append(item)
    digtsum=0
    divlst.clear()
myDic={}
myDic["Proper"]=plst
myDic["Improper"]=implst
print(myDic)
```
