```python
#Q2
class Assassin:
    total_Ass=0
    def __init__(self,name,srate):
        self.name=name
        self.srate=srate
        Assassin.total_Ass+=1
    def printDetails(self):
        print(f"Name: {self.name}\nSuccess rate: {self.srate}%\nTotal number of Assassin: {Assassin.total_Ass}")
    @classmethod
    def failureRate(cls,name,frate):
        srate=100-frate
        obj=cls(name,srate)
        return obj
    @classmethod
    def failurePercentage(cls,name,failpertg):
        srate=100-failpertg
        obj=cls(name,srate)
        return obj
john_wick = Assassin('John Wick', 100)
john_wick.printDetails()
print('================================')
nagisa = Assassin.failureRate("Nagisa", 20)
nagisa.printDetails()
print('================================')
akabane = Assassin.failurePercentage('Akabane', 10)
akabane.printDetails()
```

```python
#Q7
class Cat:
    Number_of_cats=0
    def __init__(self,clr,act):
        self.clr=clr
        self.act=act
        Cat.Number_of_cats+=1
    def changeColor(self,clr):
        self.clr=clr
    def printCat(self):
        print(f"{self.clr} cat is {self.act}")
    @classmethod
    def first_parameter(cls,clr):
        return cls(clr,act='sitting')

    @classmethod
    def no_parameter(cls):
        return cls(clr='White',act='sitting')

    @classmethod
    def second_parameter(cls,act):
        return cls('Grey',act)
print("Total number of cats:", Cat.Number_of_cats)
c1 = Cat.no_parameter()
c2 = Cat.first_parameter("Black")
c3 = Cat("Brown", "jumping")
c4 = Cat("Red", "purring")
c5 = Cat.second_parameter("playing")
print("=======================")
c1.printCat()
c2.printCat()
c3.printCat()
c4.printCat()
c5.printCat()
c1.changeColor("Blue")
c3.changeColor("Purple")
c1.printCat()
c3.printCat()
print("=======================")
print("Total number of cats:", Cat.Number_of_cats)
```

```python
#Q8
import math
class Cylinder:
    radius=5
    height=18
    def __init__(self,rad,hgt):
        print(f"Default radius={Cylinder.radius} and height={Cylinder.height}")
        Cylinder.radius=rad
        Cylinder.height=hgt
        print(f"Updated: radius={self.radius} and height={self.height}")
    @staticmethod
    def area(rad,hgt):
        print('Area:',2*math.pi*Cylinder.radius*Cylinder.height+2*math.pi*Cylinder.radius**2)
    @staticmethod
    def volume(rad,hgt):
        print('Volume:',math.pi*rad**2*hgt)
    @classmethod
    def swap(cls,hgt,rad):
        return cls(rad,hgt)
    @classmethod
    def changeFormat(cls,info):
        rad,hgt=info.split('-')
        return cls(float(rad),float(hgt))
c1 = Cylinder(0,0)
Cylinder.area(c1.radius,c1.height)
Cylinder.volume(c1.radius,c1.height)
print("===============================")
c2 = Cylinder.swap(8,3)
c2.area(c2.radius,c2.height)
c2.volume(c2.radius,c2.height)
print("===============================")
c3 = Cylinder.changeFormat("7-13")
c3.area(c3.radius,c3.height)
c3.volume(c3.radius,c3.height)
print("===============================")
Cylinder(0.3,5.56).area(Cylinder.radius,Cylinder.height)
print("===============================")
Cylinder(3,5).volume(Cylinder.radius,Cylinder.height)
```

```python
#Q10
class SultansDine:
    tot_brn=0
    tot_sell=0
    my_Dic={}
    def __init__(self,loct):
        self.loct=loct
        SultansDine.tot_brn+=1
        self.b_sell=0
        self.my_Dic={}
    def sellQuantity(self,quant):
        if quant<10:
            self.b_sell=quant*300
            SultansDine.my_Dic[self.loct]=self.b_sell
        elif quant<20:
            self.b_sell=quant*350
            SultansDine.my_Dic[self.loct]=self.b_sell
        else:
            self.b_sell=quant*400
            SultansDine.my_Dic[self.loct]=self.b_sell
        SultansDine.tot_sell+=self.b_sell
    def branchInformation(self):
        print(f"Branch Name: {self.loct}\nBranch Sell: {self.b_sell} Taka")
    @classmethod
    def details(cls):
        if SultansDine.tot_brn==0:
            print(f"Total Number of branches: {SultansDine.tot_brn}\nTotal Sell: {SultansDine.tot_sell} Taka")
        else:
            print(f"Total Number of branches: {SultansDine.tot_brn}\nTotal Sell: {SultansDine.tot_sell} Taka")
            for key,val in cls.my_Dic.items():
                print(f"Branch Name: {key}, Branch Sell: {val} Taka")
                print("Branch consists of total sell's:",'{:.2f}'.format((val/SultansDine.tot_sell)*100),"%")
SultansDine.details()
print('########################')
dhanmodi = SultansDine('Dhanmondi')
dhanmodi.sellQuantity(25)
dhanmodi.branchInformation()
print('-----------------------------------------')
SultansDine.details()
print('========================')
baily_road = SultansDine('Baily Road')
baily_road.sellQuantity(15)
baily_road.branchInformation()
print('-----------------------------------------')
SultansDine.details()
print('========================')
gulshan = SultansDine('Gulshan')
gulshan.sellQuantity(9)
gulshan.branchInformation()
print('-----------------------------------------')
SultansDine.details()
```
