```python
class TicketCounter:
    def __init__(self):
        self.pcnt=0
        self.bcnt=0
        self.tcnt=0
        self.bDic={}
        self.tDic={}
        print(f"This is a general bus and train ticket counter.")
    def servePassenger(self,*obj):
        if len(obj)==2:
            self.pass1=obj[0]
            self.fchat=obj[1]
            for k,v in fareChart.items():
                if self.pass1.tcat=='Bus Ticket' and k=='Bus':
                    self.bDic[self.pass1.name]=self.pass1.tkfee-v
                    print(f"{self.pass1.name} has bought {self.pass1.tcat}.")
                    self.pcnt+=1
                    self.bcnt+=1
                elif self.pass1.tcat=='Train Ticket' and k=='Train':
                    self.tDic[self.pass1.name]=self.pass1.tkfee-v
                    print(f"{self.pass1.name} has bought {self.pass1.tcat}.")
                    self.pcnt+=1
                    self.tcnt+=1
        elif len(obj)==3:
            self.pass1=obj[0]
            self.pass2=obj[1]
            self.fchat=obj[2]
            for k1,v1 in fareChart.items():
                if self.pass1.tcat=='Train Ticket' and k1=='Train':
                    self.tDic[self.pass1.name]=self.pass1.tkfee-v1
                    print(f"{self.pass1.name} has bought {self.pass1.tcat}.")
                    self.pcnt+=1
                    self.tcnt+=1
            for k2,v2 in fareChart.items():
                if self.pass2.tcat=='Train Ticket' and k2=='Train':
                    self.tDic[self.pass2.name]=self.pass2.tkfee-v2
                    print(f"{self.pass2.name} has bought {self.pass2.tcat}.")
                    self.pcnt+=1
                    self.tcnt+=1
                elif self.pass2.tcat==None:
                    print(f"{self.pass2.name} please buy a ticket first.")
                    break
    def showAllPassenger(self):
        print('Total passengers:',self.pcnt)
        print('Bus passengers:',self.bcnt)
        for key,val in self.bDic.items():
            print(f"Passenger Name:{key} Remaining balance:{val}") 
        print('Train passengers:',self.tcnt)
        for key1,val1 in self.tDic.items():
            print(f"Passenger Name:{key1} Remaining balance:{val1}") 
class Passenger:
    def __init__(self,*info):
        self.name=info[0]
        if len(info)==3:
            self.tcat=info[1]
            self.tkfee=info[2]
        else:
            self.tkfee=info[1]
            self.tcat=None
        if self.tcat=='Bus Ticket':
            print(f"{self.name} is waiting for Bus having {self.tkfee} taka in pocket.")
        elif self.tcat=='Train Ticket':
            print(f"{self.name} is waiting for Train having {self.tkfee} taka in pocket.")
        else:
            print(f"{self.name} has {self.tkfee} taka in pocket.Please buy a ticket.")
    def buyTicket(self,tcat):
        self.tcat=tcat
fareChart = {"Bus":550, "Train":600}
p1 = Passenger("Bob", "Bus Ticket",800)
p2 = Passenger("Simon", "Train Ticket", 950)
p3 = Passenger("David", "Train Ticket", 750)
p4 = Passenger("Carol", 1500)
print("1==============================")
tc = TicketCounter()
print("2==============================")
tc.servePassenger(p1,fareChart)
print("3==============================")
tc.servePassenger(p2,fareChart)
print("4==============================")
tc.servePassenger(p3, p4, fareChart)
print("5==============================")
p4.buyTicket("Train Ticket")
tc.servePassenger(p4,fareChart)
print("6==============================")
tc.showAllPassenger()
```
