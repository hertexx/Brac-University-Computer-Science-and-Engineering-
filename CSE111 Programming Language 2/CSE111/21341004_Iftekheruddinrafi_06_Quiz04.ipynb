```python
#Q1
class Squad:
    playerCount=0
    def __init__(self,name,squadType):
        self.name = name
        self.squadType = squadType
    def formSquad(self):
        if Squad.playerCount >= 5:
            return True
        else:
            return False
    def __str__(self):
        s = f"{self.name} is a {self.squadType} team.\n"
        s+= f"Number of players in {self.name} is {Squad.playerCount}\n"
        return s
class HogwartsQuidditchSquad(Squad):
    def __init__(self,name,squadType):
        super().__init__(name,squadType)
        self.myDic={}
    def addPlayer(self,p1,p2=None):
        if p2==None:
            self.pname1=p1[0]
            self.pposition1=p1[1]
            self.phouse1=p1[2]
            Squad.playerCount+=1
            if self.pposition1 not in self.myDic.keys():
                self.myDic[self.pposition1]=[self.pname1,self.phouse1]
            else:
                self.myDic[self.pposition1].append(self.pname1)
                self.myDic[self.pposition1].append(self.phouse1)
        else:
            self.pname1=p1[0]
            self.pposition1=p1[1]
            self.phouse1=p1[2]
            self.pname2=p2[0]
            self.pposition2=p2[1]
            self.phouse2=p2[2]
            Squad.playerCount+=2
            self.myDic[self.pposition1]=[self.pname1,self.phouse1]
            self.myDic[self.pposition2]=[self.pname2,self.phouse2]
    def __str__(self):
        s = f"{self.name} is a {self.squadType} team.\n"
        s+= f"Number of players in {self.name} is {Squad.playerCount}\n"
        s+= f"The Players are\n"
        for key,val in self.myDic.items():
            if key=='Seeker':
                s+=f"Seeker:\nPlayer name: {val[0]},House: {val[1]}"
            elif key=='Chaser':
                s+=f"\nChaser:\nPlayer name: {val[0]},House: {val[1]}"
            elif key=='Beater':
                if len(val)==2:
                    s+=f"\nBeater:\nPlayer name: {val[0]},House: {val[1]}"
                else:
                    s+=f"\nBeater:\nPlayer name: {val[0]},House: {val[1]}"
                    s+=f"\nPlayer name: {val[2]},House: {val[3]}"
            elif key=="Keeper":
                if len(val)==2:
                    s+=f"\nKeeper:\nPlayer name: {val[0]},House: {val[1]}"
                else:
                    s+=f"\nKeeper:\nPlayer name: {val[0]},House: {val[1]}"
                    s+=f"\nPlayer name: {val[2]},House: {val[3]}"
        return s
    def formSquad(self):
        if Squad.playerCount == 5:
            print(f"We have enough players to form a squad.\nBut we cannot form a perfect squad.")
        elif Squad.playerCount<5:
            print(f"We do not have enough players to form a squad.")
        else:
            print(f"We have enough players to form a squad.\nAlso we can form a perfect squad!!")
f = HogwartsQuidditchSquad("Hogwart's Dragons","Quidditch")
f.addPlayer(["Harry Potter","Seeker","Gryffindor"],["Katie Bell","Chaser","Gryffindor"])
print("1.====================================")
print(f)
print("2.====================================")
f.formSquad()
print("3.====================================")
f.addPlayer(["Vincent Crabbe","Beater","Slytherin"])
f.addPlayer(["Miles Bletchley","Keeper","Slytherin"])
print("4.====================================")
print(f)
print("5.====================================")
f.formSquad()
print("6.====================================")
f.addPlayer(["Ethan Humberstone","Keeper","Hufflepuff"])
f.formSquad()
print("7.====================================")
f.addPlayer(["Fred Weasley","Beater","Gryffindor"])
print(f)
print("8.====================================")
f.formSquad()
```
