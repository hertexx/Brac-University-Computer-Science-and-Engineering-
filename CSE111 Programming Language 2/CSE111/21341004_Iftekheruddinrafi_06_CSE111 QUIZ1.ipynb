```python
IDs = ['20201199','21121347','20141052','20341121','21241369','21272199','19241187','19101007','20101035', '21121875', '19141534', '19301552', '21141135', '21365001']
myDic={}
yearid=[]
Dic1={'CSE':[],'EEE':[],'CS':[],'Other':[]}
for items in IDs:
    if items[3:5]=='01':
        Dic1['CSE'].append(items)
    elif items[3:5]=='41':
        Dic1['CS'].append(items)
    elif items[3:5]=='21':
        Dic1['EEE'].append(items)
    else:
        Dic1['Other'].append(items)
for elem in IDs:
    if elem[:2] not in yearid:
        yearid.append(elem[:2])
for num in yearid:
    myDic[int(str(20)+num)]={}
    for key,value in Dic1.items():
        for val in value:
            if val[:2]==num:
                if key not in num:
                    myDic[int(str(20)+num)][key]=[]
                    myDic[int(str(20)+num)][key].append(val)
                else:
                    myDic[int(str(20)+num)][key].append(val)
print(myDic)
```
