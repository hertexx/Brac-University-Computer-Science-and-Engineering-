```python
#Answer
class Tree:
    count=0
    def __init__(self,tname,sdtree=False):
        self.tname=tname
        self.sdtree=sdtree
        self.myLst=[]
        Tree.count+=1
    def autumnEffect(self):
        if self.sdtree==True:
            print(self.tname,"shed their leaves in autumn")
        else:
            print(self.tname,"do not shed their leaves in autumn")
    def add_Benefit(self,*info):
        if len(info)==1:
            self.myLst.append(info[0])
        else:
            self.myLst.append(info[0])
            self.myLst.append(info[1])
    def printDetail(self):
        print("Name:",self.tname)
        print("Benefits:",end='')
        if len(self.myLst)==1:
            print(self.myLst[0],end='')
        else:
            print(self.myLst[0],end=', ')
            print(self.myLst[1],end='')
        print()
print("No. of Trees =", Tree.count)
pine = Tree('Pine')
oak = Tree('Oak', True)
maple = Tree('Maple', True)
print('###########################')
pine.autumnEffect()
oak.autumnEffect()
maple.autumnEffect()
pine.add_Benefit('Ornamental')
maple.add_Benefit('Hard-wood', 'Fruit')
oak.add_Benefit('Hard-wood', 'Shading')
print("=========================")
pine.printDetail()
print("=========================")
oak.printDetail()
print("=========================")
maple.printDetail()
print("=========================")
print('No. of Trees =', Tree.count)
```
