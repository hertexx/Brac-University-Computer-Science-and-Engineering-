```python
#10
height = int(input('Height:'))
width = int(input('Width:'))
for i in range(height):
    for j in range(width):
        if i==0 or i==height-1:
            print('*',end='')
        else:
            if i>=1 or i<height-1:
                if 1<=j<=width-2:
                    print(' ', end='')
                else:
                    print('*',end='')
    print()
```

```python
#2
import math
x=int(input('x'))
y=int(input('y'))
z=int(input('z'))
a=math.acos((y**2+z**2-x**2)/2*y*z)
b=math.acos((z**2+z**2-y**2)/2*x*z)
c=math.acos((y**2+x**2-z**2)/2*y*x)
if 0<=(a+b+c)<=math.pi:
    print('v')
else:
    print('nv')
```
