```python
#1
inp = open('input1.txt', 'r')
out = open('output1.txt', 'w')
n = int(inp.readline())
lst = list(map(int,inp.readline().split()))
def merge(a, b):
  # write your code here
  # Here a and b are two sorted list
  # merge function will return a sorted list after merging a and b
  lst1 = a
  n = len(a)
  lst2 = b
  m = len(b)
  a = []
  i, j = 0, 0
  while True:
    if i == n:
        a += lst2[j:]
        break
    elif j == m:
        a += lst1[i:]
        break
    elif lst1[i] < lst2[j]:
        a.append(lst1[i])
        i += 1
    elif lst1[i] > lst2[j]:
        a.append(lst2[j])
        j += 1
    elif lst1[i] == lst2[j]:
        a.append(lst1[i])
        a.append(lst2[j])
        i += 1
        j += 1

  return a
def mergeSort(arr):
  if len(arr) <= 1:
    return arr
  else:
    mid = len(arr)//2
    a1 = mergeSort(arr[:mid])  # write the parameter
    a2 = mergeSort(arr[mid:])  # write the parameter
    return merge(a1, a2)
a = mergeSort(lst)
for i in a:
    out.writelines(str(i)+" ")
inp.close()
out.close()
```
