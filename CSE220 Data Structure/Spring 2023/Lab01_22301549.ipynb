```python
from re import I
# Complete the functions defined in this cell


# Test 01: Play Right
def playRight(sequence,beats):
  for elem in beats:
    if elem==1:
      i=len(sequence)-1
      temp=None
      while i>=0:
        if i==len(sequence)-1:
          temp=sequence[len(sequence)-1]
        sequence[i]=sequence[i-1]
        i-=1
      sequence[0]=temp
  return sequence
# Test 02: Discard Cards
def discardCards(cards,number):
  i=0                          # [1,2,3,2,8,2,2,5,7]
  while i<len(cards):
    if cards[i]==number:
      for j in range(i,len(cards)-1):
        cards[j]=cards[j+1]
      cards[len(cards)-1]=0
    if cards[i]==number:
      for j in range(i,len(cards)-1):
        cards[j]=cards[j+1]
    i+=1
  return cards
#Test 03: Merge Lineup
def mergeLineup(pokemon_1, pokemon_2):
  a=np.array([0]*(len(pokemon_1)))
  i=0
  j=len(pokemon_2)-1
  while j>=0:
    x=0
    y=0
    if pokemon_1[i]!=None:
      x=pokemon_1[i]
    if pokemon_2[j]!=None:
      y=pokemon_2[j]
    a[i]=x+y
    i+=1
    j-=1
  return a
# Test 04: Balance your Salami
def balanceSalami(salami):
  flag=False
  sum=0
  i=0
  while i<len(salami):
    sum+=salami[i]
    i+=1
  i=0
  sum1=0
  while i<len(salami):
    sum-=salami[i]
    sum1+=salami[i]
    if sum1==sum:
      flag=True
      return flag
    i+=1
  return flag
# Test 05: Protecc Salami
def protectSalami(salami):
  newArr=np.array([0]*len(salami))
  cnt=0
  temp=salami
  flag=False
  for i in salami:
    if i in newArr:
      cnt+=1
    else:
      newArr[i]+=1
  for i in range(len(newArr)-2):
    for j in range(i+1,len(newArr)-1):
      if newArr[i]>1 and newArr[j]>1 and newArr[i]==newArr[j]:
        flag=True
  return flag


# Test 05: Odd Even Wave
def waveYourFlag(arr):                     # 45 23 78 84 41
  if arr[0]%2==0:
    i=1
    while i<len(arr):
      if arr[i]%2==0 and arr[i+1]!=0:
        temp=arr[i]
        arr[i]=arr[i+1]
        arr[i+1]=temp
      i+=2
  else:
    i=1
    while i<len(arr):
      if arr[i]%2!=0 and arr[i+1]%2==0:
        temp=arr[i]
        arr[i]=arr[i+1]
        arr[i+1]=temp
      i+=2
  return arr

```
