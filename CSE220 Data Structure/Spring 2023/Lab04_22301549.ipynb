```python
class Patient:
  def __init__(self,elem=None,next=None,prev=None):
    self.elem,self.next,self.prev=elem,next,prev
class WRM:
  def __init__(self):
    self.dh=Patient()
    self.dh.next=self.dh
    self.dh.prev=self.dh
  def RegisterPatient(self,id,name,age,bloodgroup):
    n=Patient([id,name,age,bloodgroup])
    head=self.dh.next
    if head.elem==None:
      self.dh.next=n
      n.prev=self.dh
      n.next=self.dh
    else:
      while head.next.elem!=None:
        head=head.next
      head.next=n
      n.prev=head
      n.next=self.dh
  def ServePatient(self):
    head=self.dh.next
    if head.elem==None:
      print("\nNo Patient To Serve.")
    else:
      pre=self.dh.next.next
      self.dh.next=self.dh.next.next
      pre.prev=self.dh
      print(f"\nServing The Patient Name: {head.elem[1]}.")
      head.next=None
      head.prev=None
      head.elem=None
  def CancelAll(self):
    #self.dh.next=None
    #self.dh.prev=None
    if self.dh.next.elem==None:
      print("\nNo Appointments to Cancel.")
    else:
      head=self.dh.next
      while head.elem!=None:
        pre=self.dh.next.next
        self.dh.next=self.dh.next.next
        pre.prev=self.dh
        del_elem=head
        head=head.next
        del_elem.next=None
        del_elem.prev=None
        del_elem.elem=None
      print("\nSuccessfully All Appointments Cancelled.")
  def CanDoctorGoHome(self):
    head=self.dh.next
    if head.elem==None:
      return True
    else:
      return False
  def ShowAllPatient(self):
    head=self.dh.next
    if head.elem==None:
       print(f"\nNo Patient Waiting.")
    else:
      while head.elem!=None:
        print(head.elem[0],end=' ')
        head=head.next
```
