```python
#10
def jump(a):
  cnt=0
  cnt1=0
  while cnt1<len(a)-1:
    cnt1+=a[cnt1]
    cnt+=1
  if cnt==0:
    return -1
  else:
    return cnt
jump([1,4,3,2,6,7])
```

```python
#11
import math
def maxminfun(a):
  max=a[0]
  min=a[0]
  for i in range(len(a)):
    if a[i]>max:
      max=a[i]
    if a[i]<min:
      min=a[i]
  return [max,min]
def height(K,N,Arr):
  avg=math.ceil((maxminfun(Arr)[0]-maxminfun(Arr)[1])/2)
  a=[]
  for i in range(len(Arr)):
    if Arr[i]>avg:
      a+=[Arr[i]-K]
    elif Arr[i]<=avg:
      a+=[Arr[i]+K]
  return (maxminfun(a)[0]-maxminfun(a)[1])
print(height(2,4,[1, 5, 8, 10]))
print(height(3,5,[3, 9, 12, 16, 20]))
```

```python
#12
def platform(n,arr,dep):
  plcnt=1
  train=0
  for i in range(arr[0],dep[-1],1):
    for j in range(n):
      if i==arr[j]:
        train+=1
        if train>plcnt:
          plcnt+=1
      if i==dep[j]:
        train-=1
  return plcnt
print(platform(6,[900, 940, 950, 1100, 1500, 1800],[910, 1200, 1120, 1130, 1900, 2000]))
print(platform(3,[900, 1100, 1235],[1000, 1200, 1240]))
```

```python
#13
def findMinDiff(arr, n, m):
  if (m == 0 or n == 0):
    return 0
    arr.sort()
    if (n < m):
        return -1
    min_diff = arr[n-1] - arr[0]
    for i in range(len(arr) - m + 1):
        min_diff = min(min_diff,  arr[i + m - 1] - arr[i])
    return min_diff
arr = [12, 4, 7, 9, 2, 23, 25, 41,
      30, 40, 28, 42, 30, 44, 48,
      43, 50]
m = 7  # Number of students
n = len(arr)
print("Minimum difference is", findMinDiff(arr, n, m))
```

```python
#14
def con(n,arr):
  cnt1=0
  a=arr
  cnt2=1
  a2=[]
  while cnt1<len(arr)-1:
    a1=[a[cnt1]]
    for i in range(len(arr)):
      if a[cnt1]+cnt2 in a:
        a1+=[a[cnt1]+cnt2]
      else:
        break
      cnt2+=1
    cnt2=1
    cnt1+=1
    if max(len(a1),len(a2))==len(a1):
      a2=a1
  return len(a2)
print(con(3,[2,6,1,9,4,5,3]))
print(con(3,[1,9,3,10,4,20,2]))
```

```python
#15
def med(a1,a2):
  a=a1
  for i in range(len(a2)-1):
    a.append(a2[i])
  for j in range(len(a)):
    for k in range(len(a)-j-1):
      if a[j]>a[j+1]:
        a[j],a[j+1]=a[j+1],a[j]
  b=(len(a)-1)/2
  if type(b)==type(float()):
    return (a[round(b)]+a[round(b)+1])/2
  else:
    return a[round(b)]
med([1,2],[3,4])
```
