```python
#Assignment Part 1
class Patient:
  def __init__(self,elem=None, next = None, prev = None):
    self.elem,self.next,self.prev=elem,next,prev
```

```python
class WRM:
  def __init__(self):
    self.dh = Patient()
    self.dh.next = self.dh
    self.dh.prev = self.dh
  def registerPatient(self,id, name, age, bloodgroup):
    n = Patient([id, name, age, bloodgroup], self.dh)
    tail = self.dh.prev
    tail.next = n
    self.dh.prev = n
    n.prev = tail
  def servePatient(self):
    if self.dh.next==self.dh:
      print("\nNo Patient To Serve.")
    else:
      print(f"\nServing The Patient Name: {self.dh.next.elem[1]}.")
      self.dh.next = self.dh.next.next
      self.dh.next.prev = self.dh.next
  def showAllPatient(self):
    if self.dh.next==self.dh:
       print(f"\nNo Patient Waiting.")
    else:
      head = self.dh.next
      while head!=self.dh:
        print(head.elem[1],end=' ')
        head = head.next
  def canDoctorGoHome(self):
    if self.dh.next == self.dh:
      return True
    return False
  def cancelAll(self):
    if self.dh.next == self.dh:
      print("\nNo Appointments to Cancel.")
    else:
      self.dh.next = self.dh
      self.dh.prev = self.dh
      print("\nSuccessfully All Appointments Cancelled.")
  def ReverseTheLine(self):
    if self.dh.next == self.dh:
      print("\nReverseing Not Possible.")
    else:
      tail = self.dh.prev
      newWRM = Patient()
      newWRM.next = newWRM
      newWRM.prev = newWRM
      while tail!= self.dh:
        n = Patient(tail.elem,newWRM)
        t2 = newWRM.prev
        t2.next = n
        newWRM.prev = n
        n.prev = t2
        tail = tail.prev
      self.dh = newWRM
      print("\nSuccessfully Reverseing Completed.")
```

**Assignment Part 2: Stack**

Linked List based Stack is implemented in the following cell.

```python
class Node:
  def __init__(self,elem=None,next=None):
    self.elem = elem
    self.next = next

class Stack:
  def __init__(self):
    self.__top = None

  def push(self,elem):
    nn = Node(elem,self.__top)
    self.__top = nn

  def pop(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    e = self.__top
    self.__top = self.__top.next
    return e.elem

  def peek(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    return self.__top.elem

  def isEmpty(self):
    return self.__top == None
```

Task 1: Diamond Count

```python
def diamond_count(stack,string):
  #TO DO
  cnt = 0
  for i in range(len(string)):
    if string[i]=='<':
      stack.push(string[i])
    elif stack.isEmpty()!= True:
      if string[i] == '>':
        stack.pop()
        cnt+=1
  return cnt

print('Test 01')
stack = Stack()
string = '<..><.<..>> '
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')


print('Test 02')
stack = Stack()
string = '<<<..<......<<<<....>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 1
unittest.output_test(returned_value, 1)
print('-----------------------------------------')


print('Test 03')
stack = Stack()
string = '>>><...<<..>>...>...>>>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')
```

Task 2: Tower of Blocks

```python
def remove_block(stack, n):
  nst = Stack()
  cnt=1
  while stack.isEmpty()!= True:
    if cnt!=n:
      nst.push(stack.pop())
    else:
      stack.pop()
    cnt+=1
  while nst.isEmpty()!= True:
    stack.push(nst.pop())
  return stack

print('Test 01')
st = Stack()
st.push(4)
st.push(19)
st.push(23)
st.push(17)
st.push(5)
print('Stack:')
print_stack(st)
print('------')
remove_block(st,2)
print('After Removal')
print_stack(st)
print('------')

print()
print('======================================')
print()

print('Test 02')
st = Stack()
st.push(73)
st.push(85)
st.push(15)
st.push(41)
print('Stack:')
print_stack(st)
print('------')
remove_block(st,3)
print('After Removal')
print_stack(st)
print('------')

print()
print('======================================')
print()
```

```python
def conditional_reverse(stack):
  revStack = Stack()
  i = None
  while st.isEmpty()!= True:
    j = st.pop()
    if i!=j:
      revStack.push(j)
    i=j
  return revStack
print('Test 01')
st=Stack()
st.push(10)
st.push(10)
st.push(20)
st.push(20)
st.push(30)
st.push(10)
st.push(50)
print('Stack:')
print_stack(st)
print('------')
reversed_stack=conditional_reverse(st)
print('Conditional Reversed Stack:')
print_stack(reversed_stack) # This stack contains 50, 10, 30, 20, 10 in this order whereas top element should be 10
print('------')
```
