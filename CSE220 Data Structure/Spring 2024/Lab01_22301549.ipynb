```python
#Task 01: Merge Lineup
def mergeLineup(pokemon_1, pokemon_2):
  result = np.array([0]*len(pokemon_1))
  i = 0
  j = len(pokemon_1)-1
  while j>-1:
    if pokemon_1[i]==None:
      result[i]= 0 +pokemon_2[j]
    elif pokemon_2[j]==None:
      result[i]= pokemon_1[i]+ 0
    elif pokemon_2[j]==None and pokemon_1[i]==None:
      result[i]= 0
    else:
      result[i]= pokemon_1[i]+pokemon_2[j]
    i+=1
    j-=1
  return result
print("///  Task 01: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))
pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
from pickle import FALSE
# Task 02: Discard Cards
def discardCards(cards, t):
  tag = True
  stime = True
  i = 0
  j = 0
  while i<len(cards)-1:
    if cards[i]!=t and stime:
      i+=1
      j+=1
    elif cards[i]==t or stime==False:
      stime= False
      if tag and cards[i]==t:
        tag = False
        j=i
        while j<len(cards)-1:
          cards[j]=cards[j+1]
          j+=1
      elif cards[i]==t:
        tag = True
        i+=1
      elif cards[i]!=t:
        i+=1
  return cards
print("///  Task 02: Discard Cards  ///")
cards = np.array([1,3,7,2,5,2,2,2,0])
returned_value = discardCards(cards, 2)
print(f'Task 2: {returned_value}') # This should print [1,3,7,5,2,2,0,0,0]
unittest.output_test(returned_value, np.array([1,3,7,5,2,2,0,0,0]))

cards = np.array([5,5,5,0,0])
returned_value = discardCards(cards, 5)
print(f'Task 2: {returned_value}') # This should print [5,0,0,0,0]
unittest.output_test(returned_value, np.array([5,0,0,0,0]))
```

```python
# Task 03: DUBER Fare Splitting
def findGroups(money, fare):
  # TO DO
  # Print outputs inside the method
  cnt=1
  for i in range(len(money)):
    for j in range(len(money)):
      if money[j]==0:
        continue
      elif money[i]==fare:
        print(f"Group {cnt}: {money[i]}")
        cnt+=1
        money[i]=0
        break
      elif money[i]+money[j]==fare:
        print(f"Group {cnt}: {money[i]}, {money[j]}")
        cnt+=1
        money[i]=0
        money[j]=0
  print("Ungrouped: ",end="")
  for k in money:
    if k!=0:
      print(k,end=' ')
print("///  Task 03: DUBER Fare Splitting  ///")
money = np.array( [120, 100, 150, 50, 30])
fare = 150
print(f'Task 3:')
findGroups(money, fare) # This should print

# Group 1 : 120, 30
# Group 2 : 100, 50
# Group 3 : 150


money = np.array( [60, 150, 60, 30, 120, 30])
fare = 180
print(f'Task 3:')
findGroups(money, fare) # This should print

# Group 1 : 60, 120
# Group 2 : 30, 150
# Ungrouped : 30 60
```

```python
import math
arr=[10, 8, 13, 9, 14, 25, -5, 20, 7, 7, 4]
sum=0
for i in range(len(arr)):
  sum+=arr[i]
mean=sum/len(arr)
print("The mean of the numbers is: %.12f" %mean)
x=0
for j in range(len(arr)):
  x=x+ (arr[j]-mean)**2
y=x/(len(arr)-1)
sd=math.sqrt(y)
print("The standard deviation is: %.2f"%sd)
cnt=0
for l in range(len(arr)):
  if mean-1.5*sd>=arr[l] or arr[l]>=mean+1.5*sd:
    cnt+=1
newArr=np.array([0]*cnt)
cnt1=0
for k in range(len(arr)):
  if mean-1.5*sd>=arr[k] or arr[k]>=mean+1.5*sd:
    newArr[cnt1]=arr[k]
    cnt1+=1
print('New array:',newArr)
```
