```python
class Node_pair:
  def __init__(self, key, value, next = None):
    self.key, self.value, self.next = key, value, next


class Hashtable:
  def __init__(self, size):
    self.ht = [None]*size


  def insert(self, s):
    if self.search_hashtable(s) == 'Found':
      print(s,'Already Inserted. Cannot reinsert.')
      print('===============================')
      return
    hashed_index = self.__hash_function(s[0])
    pair = Node_pair(s[0], s[1])
    if self.ht[hashed_index] == None:
      self.ht[hashed_index] = pair
    else:
      pair.next = self.ht[hashed_index]
      self.ht[hashed_index] = pair


  def create_from_array(self, arr):
    for i in arr:
      self.insert(i)

  def print_hashtable(self):
    idx = 0
    for i in self.ht:
      print(idx, ':', end = ' ')
      head = i
      while head != None:
        print(f'(key: {head.key}, value: {head.value})', end = '-->')
        head = head.next
      print('None')
      print()
      idx += 1
  def __hash_function(self, key):
    s=key
    if len(s)%2!=0:
      s=s+'N'
    list_1=[None]*int(len(s)/2)
    index=0
    while len(s)!=0:
      value=str(ord(s[0]))+str(ord(s[1]))
      list_1[index]=value
      index+=1
      s=s[2:]
    total=0
    for i in list_1:
      total=total+int(i)
    return total%len(self.ht)

  def search_hashtable(self, s):
    idx = self.__hash_function(s[0])
    temp=self.ht[idx]
    while temp!=None:
      if temp.key==s[0]:
        return 'Found'
      temp=temp.next
    return "Not Found"
```

```python
class Node:
  def __init__(self, value=None, next = None):
    self.value = value
    self.next = next

class HashTable:
  def __init__(self, length):
    n = Node()
    self.ht = [None] * length
    self.length = length

  def show(self):
    count = 0
    for i in self.ht:
      temp = i
      print(count, end=" ")
      while temp!=None:
        print (temp.value, end="-->")
        temp = temp.next
      count += 1
      print()
  def __hash_function(self, key):
    s1 = 0
    s2 = 0
    for i in range(len(key)):
      if i%2!=0:
        s1 += ord(key[i])
      else:
        s2 += ord(key[i])

    if len(key)%2!=0:
      return s1%self.length
    return s2%self.length
  def insert(self, key, value):
    k = self.__hash_function(key)
    if self.ht[k]==None:
      n = Node((key, value))
      self.ht[k] = n
    else:
      if self.ht[k].value[1]<=value:
        n = Node((key, value),self.ht[k])
        self.ht[k] = n
      else:
        temp = self.ht[k]
        while temp.next!=None:
          if temp.next.value[1]<value:
            n = Node((key, value), temp.next)
            temp.next = n
            break
          temp= temp.next
        if temp.next==None:
          n = Node((key, value))
          temp.next = n
```

```python
class Node_pair:
  def __init__(self, key, value, next = None):
    self.key, self.value, self.next = key, value, next


class Hashtable:
  def __init__(self, size):
    self.ht = [None]*size

  def insert(self, s):
    hashed_index = self.__hash_function(s[0])
    pair = Node_pair(s[0], s[1])
    if self.ht[hashed_index] == None:
      self.ht[hashed_index] = pair
    else:
      pair.next = self.ht[hashed_index]
      self.ht[hashed_index] = pair


  def create_from_array(self, arr):
    for i in arr:
      self.insert(i)

  def print_hashtable(self):
    idx = 0
    for i in self.ht:
      print(idx, ':', end = ' ')
      head = i
      while head != None:
        print(f'({head.key}, {head.value})', end = '-->')
        head = head.next
      print('None')
      idx += 1

  def __hash_function(self, key):
    return (key+3)%6
  def remove(self, key):
    k = self.__hash_function(key)
    temp = self.ht[k]
    if temp.key == key and temp.next!=None:
      self.ht[k] = self.ht[k].next
    elif temp.key == key and temp.next==None:
      self.ht[k] = None
    else:
      while temp.next!=None:
        if temp.next.key == key and temp.next.next!=None:
          temp.next = temp.next.next
        elif temp.next.key == key and temp.next.key == key:
          temp.next = None
        temp = temp.next
```
