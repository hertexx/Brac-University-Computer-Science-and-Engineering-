```python
import math
import random

def minmaxAB(dep, is_maxing, alpha, beta, maxVal, minVal):
    if dep == 5:
      radVal = random.randint(1, 10)
      stgMax = math.log2(maxVal + 1) + (maxVal / 10)
      stgMin = math.log2(minVal + 1) + (minVal / 10)
      a = random.choice([0, 1])
      return stgMax  - stgMin + ((-1) ** a) * (radVal / 10)
    if not is_maxing:
        minEval = float('inf')
        for m in range(2):
            ev = minmaxAB(dep + 1, True, alpha, beta, maxVal, minVal)
            minEval = min(minEval, ev)
            beta = min(beta, ev)
            if beta <= alpha:
                break
        return minEval
    else:
        maxEval = float('-inf')
        for m in range(2):
            ev = minmaxAB(dep + 1, False, alpha, beta, maxVal, minVal)
            maxEval = max(maxEval, ev)
            alpha = max(alpha, ev)
            if beta <= alpha:
                break
        return maxEval
stPlayer = int(input("Enter starting player for game 1 (0 for Carlsen, 1 for Caruana): "))
mcStrength = float(input("Enter base strength for Carlsen: "))
fcStrength = float(input("Enter base strength for Caruana: "))

print('=================================================================')
mcWins, fcWins, draws = 0, 0, 0
mResult = "Draw"
for g in range(4):
    isMaxin = (g % 2 == 0)
    maxV, minV = (mcStrength, fcStrength) if (stPlayer == 0) == isMaxin else (fcStrength, mcStrength)
    u = minmaxAB(0, True, float('-inf'), float('inf'), maxV, minV)

    ouput = "Draw"
    if u > 0:
        ouput = "Magnus Carlsen (Max)" if isMaxin else "Magnus Carlsen (Min)"
        mcWins += 1
    elif u < 0:
        ouput = "Fabiano Caruana (Max)" if isMaxin else "Fabiano Caruana (Min)"
        fcWins += 1
    else:
        draws += 1
    print(f"Game {g + 1} Winner: {ouput} (Utility value: {u:.2f})")
    stPlayer = 1 - stPlayer

if mcWins > fcWins:
    mResult = "Magnus Carlsen"
elif fcWins > mcWins:
    mResult = "Fabiano Caruana"

print("\nOverall Results:")
print(f"Magnus Carlsen Wins: {mcWins}")
print(f"Fabiano Caruana Wins: {fcWins}")
print(f"Draws: {draws}")
print(f"Overall Winner: {mResult}")
```

```python
import math
import random

def minimaxAB(d, isMaxing, alpha, beta, maxVal, minVal, mindControl=False):
    if d == 5:
      radVal = random.randint(1, 10)
      stgMax = math.log2(maxVal + 1) + (maxVal / 10)
      stgMin = math.log2(minVal + 1) + (minVal / 10)
      a = random.choice([0, 1])
      return stgMax  - stgMin + ((-1) ** a) * (radVal / 10)

    if not isMaxing:
        if mindControl:
            minEval = float('-inf')
            for m in range(2):
                e = minimaxAB(d + 1, True, alpha, beta, maxVal, minVal, mindControl)
                minEval = max(minEval, e)
        else:
          minEval = float('inf')
          for m in range(2):
            e = minimaxAB(d + 1, True, alpha, beta, maxVal, minVal, mindControl)
            minEval = min(minEval, e)
            beta = min(beta, minEval)
            if beta <= alpha:
                break
        return minEval
    else:
        maxEval = float('-inf')
        for m in range(2):
            e = minimaxAB(d + 1, False, alpha, beta, maxVal, minVal, mindControl)
            maxEval = max(maxEval, e)
            alpha = max(alpha, maxEval)
            if beta <= alpha:
                break
        return maxEval

stPlayer = int(input("Enter who goes first (0 for Light, 1 for L): "))
mcCost = float(input("Enter the cost of using Mind Control: "))
lStg = float(input("Enter base strength for Light: "))
ligStg = float(input("Enter base strength for L: "))

print('=================================================================')

maxVal, minVal = (ligStg, lStg) if stPlayer == 0 else (lStg, ligStg)

mvalue = minimaxAB(0, True, float('-inf'), float('inf'), maxVal, minVal, False)
minimaxMCAB = minimaxAB(0, True, float('-inf'), float('inf'), maxVal, minVal, True)
mAftercost = minimaxMCAB - mcCost

d = "NOT use" if mvalue >= mAftercost else "use"

print(f"Minimax value without Mind Control: {mvalue:.2f}"),
print(f"Minimax value with Mind Control: {minimaxMCAB:.2f}"),
print(f"Minimax value with Mind Control after incurring the cost: {mAftercost:.2f}"),
print(f"Light should {d} Mind Control.")
```
