# ***Dataset Pre-processing***

```python
# For numerical features - use mean imputation
for col in ['Size_sqft', 'Num_Bedrooms', 'Num_Bathrooms', 'Floor_Number',
            'Building_Age_Years', 'Distance_to_CityCenter_km']:
    dataset[col] = dataset[col].fillna(dataset[col].median())
```

```python
print("Missing values in each column:")
print(dataset.isnull().sum())
```

```python
for col in ['Location', 'Has_Balcony', 'Parking_Available', 'Nearby_Schools', 'Security_Level']:
    dataset[col] = dataset[col].fillna(dataset[col].mode()[0])
```

```python
print("Missing values in each column:")
print(dataset.isnull().sum())
```

```python
from sklearn.preprocessing import OneHotEncoder

# Binary encoding for binary features
binary_mapping = {'No': 0, 'Yes': 1}
dataset['Has_Balcony'] = dataset['Has_Balcony'].map(binary_mapping)
dataset['Parking_Available'] = dataset['Parking_Available'].map(binary_mapping)
dataset['Nearby_Schools'] = dataset['Nearby_Schools'].map({'Few': 0, 'Many': 1})

# Ordinal encoding for Security_Level
security_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
dataset['Security_Level'] = dataset['Security_Level'].map(security_mapping)
location_dummies = pd.get_dummies(dataset['Location'], prefix='Location', drop_first=True)
dataset = pd.concat([dataset, location_dummies], axis=1)
```

```python
dataset = dataset.drop(columns=['Location', 'Price_Category'])
```

```python
dataset
```

# ***Normalisation and Dataset Splitting***

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

# Assume dataset is already loaded and preprocessed with missing values handled and categorical features encoded
# This picks up where your preprocessing left off

# 1. Identify features to scale
# Numerical features
numerical_features = ['Size_sqft', 'Num_Bedrooms', 'Num_Bathrooms', 'Floor_Number',
                     'Building_Age_Years', 'Distance_to_CityCenter_km']

# Ordinal encoded categorical features
ordinal_features = ['Security_Level']

# Combine features that need scaling
features_to_scale = numerical_features + ordinal_features

# Binary features (already in 0-1 range, don't need scaling)
binary_features = ['Has_Balcony', 'Parking_Available', 'Nearby_Schools']

# 2. Split dataset into features and target
X = dataset.drop('Price_Category_Encoding', axis=1)
y = dataset['Price_Category_Encoding']

# 3. Split into training and testing sets (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y  # Maintain class distribution in both sets
)

# 4. Initialize scaler
scaler = MinMaxScaler()

# 5. Fit scaler on training data only (to avoid data leakage)
scaler.fit(X_train[features_to_scale])

print("per-feature minimum before scaling:\n {}".format(X_train.min(axis=0)))
print("")
print("per-feature maximum before scaling:\n {}".format(X_train.max(axis=0)))
print("")

# 6. Transform both training and test data
X_train[features_to_scale] = scaler.transform(X_train[features_to_scale])
X_test[features_to_scale] = scaler.transform(X_test[features_to_scale])

print("per-feature minimum before scaling:\n {}".format(X_train.min(axis=0)))
print("")
print("per-feature maximum before scaling:\n {}".format(X_train.max(axis=0)))
```

# ***Model Training & Testing***

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, classification_report, roc_auc_score, RocCurveDisplay
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import label_binarize
import numpy as np

# Define models
models = {
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Neural Network": MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=1000, random_state=42)
}

# Dictionary to store results
results = {
    'Accuracy': [],
    'Precision': [],
    'Recall': [],
    'AUC Score': []
}

# Binarize y_test for AUC-ROC calculation
y_test_binarized = label_binarize(y_test, classes=[0, 1, 2])

# Train and evaluate models
for name, model in models.items():
    # Training
    model.fit(X_train, y_train)

    # Predictions
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test) if hasattr(model, "predict_proba") else None

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')

    # Store results
    results['Accuracy'].append(accuracy)
    results['Precision'].append(precision)
    results['Recall'].append(recall)

    # AUC-ROC (One-vs-Rest)
    if y_proba is not None:
        auc = roc_auc_score(y_test_binarized, y_proba, multi_class='ovr')
        results['AUC Score'].append(auc)
    else:
        results['AUC Score'].append(None)

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Low', 'Medium', 'High'], yticklabels=['Low', 'Medium', 'High'])
    plt.title(f'{name} Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

    # Classification Report
    print(f'\n{name} Classification Report:')
    print(classification_report(y_test, y_pred))

# Model Comparison Bar Charts
plt.figure(figsize=(10, 6))
x = np.arange(len(models))
width = 0.2

plt.bar(x - width, results['Accuracy'], width, label='Accuracy')
plt.bar(x, results['Precision'], width, label='Precision')
plt.bar(x + width, results['Recall'], width, label='Recall')

plt.xticks(x, models.keys())
plt.ylabel('Score')
plt.title('Model Comparison: Accuracy, Precision, Recall')
plt.legend()
plt.ylim(0, 1)
plt.show()

# ROC Curve for each model (One-vs-Rest)
plt.figure(figsize=(10, 8))
for name, model in models.items():
    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)
        for class_id in range(3):
            RocCurveDisplay.from_predictions(
                y_test_binarized[:, class_id],
                y_proba[:, class_id],
                name=f"{name} (Class {['Low', 'Medium', 'High'][class_id]})",
                plot_chance_level=(class_id==0)
            )
plt.title('ROC Curves (One-vs-Rest)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

# Print AUC Scores
print("\nAUC Scores (One-vs-Rest):")
for name, auc in zip(models.keys(), results['AUC Score']):
    print(f"{name}: {auc:.3f}" if auc else f"{name}: Not Available")
```
