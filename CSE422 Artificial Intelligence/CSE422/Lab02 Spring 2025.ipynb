```python
import random
import json
def fitnessSimulationTrading(histPrice, strat, iniCap):
    cap = iniCap
    for pChange in histPrice:
        tSize = strat["trade_size"] / 100 * cap
        if pChange < -strat["stop_loss"]:
            loss = tSize * (strat["stop_loss"] / 100)
            cap -= loss
        elif pChange > strat["take_profit"]:
            profit = tSize * (strat["take_profit"] / 100)
            cap += profit
        else:
            cap += tSize * (pChange / 100)
    return cap - iniCap

def ChromoCrossover(p1, p2):
    cPoint = random.randint(0, 2)
    c1, c2 = p1.copy(), p2.copy()
    if cPoint == 0:
        c1["stop_loss"], c2["stop_loss"] = p2["stop_loss"], p1["stop_loss"]
    elif cPoint == 1:
        c1["take_profit"], c2["take_profit"] = p2["take_profit"], p1["take_profit"]
    else:
        c1["trade_size"], c2["trade_size"] = p2["trade_size"], p1["trade_size"]
    return c1, c2

# Function to perform mutation on a chromosome
def ChromoMutation(chromo):
    mRate = 0.05
    if random.random() < mRate:
        mGene = random.choice(["stop_loss", "take_profit", "trade_size"])
        mValue = random.uniform(1, 10)
        chromo[mGene] = mValue
    return chromo

# fun evolve e pop
def evoPop(pop, histPrices, gens, iniCap, elitism=2):
    for g in range(gens):
        #fitness of pop
        pop.sort(key=lambda x: fitnessSimulationTrading(histPrices, x, iniCap), reverse=True)

        # Elitism: Retain the top-performing chromosomes
        newPop = pop[:elitism]

        # Crossover: Create offspring from the remaining chromosomes
        while len(newPop) < len(pop):
            parent1, parent2 = random.sample(pop[:len(pop)//2], 2)
            child1, child2 = ChromoCrossover(parent1, parent2)
            newPop.append(ChromoMutation(child1))
            newPop.append(ChromoMutation(child2))
        pop = newPop

    # Return the best chromosome
    BStrategy = max(pop, key=lambda x: fitnessSimulationTrading(histPrices, x, iniCap))
    return BStrategy, fitnessSimulationTrading(histPrices, BStrategy, iniCap)


iniCap = int(input('"Capital to Start With”: $'))
hisPrices = list(map(float,input().split(',')[1:-1]))
inipop = json.loads(input('"initial_population": '))
gens = int(input('"generations": '))
BStrategy, finalProfit = evoPop(inipop, hisPrices, gens, iniCap) #evolve_population

print("Best Strategy:", BStrategy,',')
print("Final Profit:", finalProfit)
```
