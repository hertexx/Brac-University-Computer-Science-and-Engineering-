```python
import heapq
def astarSearch(st,end):
    ini ={st:None}   #parent initial
    dis ={st: 0}
    q = []
    heapq.heappush(q,(heur_val[st],st))
    while q:
        #print(q)
        t = heapq.heappop(q)
        #print(t)
        for element in g[t[1]]:
            #print(dis)
            #print(element)
            td = dis[t[1]]+ element[1]
            if element[0] not in dis or td<dis[element[0]]:
                dis[element[0]] = td
                as_cost=dis[element[0]]+ heur_val[element[0]]
                heapq.heappush(q,(as_cost,element[0]))
                ini[element[0]] = t[1]
                #print(ini)
            #print('==')
        #print(ini)
    outputPrint(ini,st,end,dis)

def outputPrint(ini,st,e,dis):
    ph=[]
    ph.append(e)
    loc_elem = e
    while loc_elem is not None:
        ph.append(ini[loc_elem])
        loc_elem = ini[loc_elem]
    ph.reverse()
    if ph is not None:
        with open("lab1o.txt", 'w') as f:
            f.write(f"Path: {' -> '.join(ph[1:])}\n")
            f.write(f"Total distance: {dis[e]} km\n")
    else:
        with open("lab1o.txt", 'w') as f:
            f.write("NO PATH FOUND\n")

inp=open("lab1inp.txt","r")
heur_val={}
g={}
for elem in inp:
    n = elem.split(" ")
    heur_val[n[0]] = int(n[1])
    g[n[0]] = []
    for e in range(2,len(n)-1,2):
        cities = n[e]
        dist = n[e+1]
        g[n[0]].append((cities,int(dist)))
print(heur_val,g)
st = input("Start node: ")
end = input("Destination: ")

astarSearch(st,end)
```
